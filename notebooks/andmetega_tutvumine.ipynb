```python
data = pd.read_csv('toorandmed_aasta.csv')
df_aasta = data
df_full = pd.read_csv('toorandmed.csv')

# Basic summaries
print("toorandmed_aasta.csv shape:", df_aasta.shape)
print("toorandmed.csv       shape:", df_full.shape)
print()

# Columns
cols_a = set(df_aasta.columns)
cols_f = set(df_full.columns)
print("Columns only in toorandmed_aasta.csv:", sorted(cols_a - cols_f))
print("Columns only in toorandmed.csv:", sorted(cols_f - cols_a))
print("Common columns count:", len(cols_a & cols_f))
print()

# Dtype differences for common columns
common_cols = [c for c in df_full.columns if c in df_aasta.columns]
dtype_diffs = {c: (df_full[c].dtype, df_aasta[c].dtype) for c in common_cols if df_full[c].dtype != df_aasta[c].dtype}
print("Dtype differences (column: (toorandmed, toorandmed_aasta)):")
print(dtype_diffs)
print()

# Row/key differences using course_uuid (if present)
if 'course_uuid' in df_full.columns and 'course_uuid' in df_aasta.columns:
    ids_f = set(df_full['course_uuid'])
    ids_a = set(df_aasta['course_uuid'])
    only_in_a = ids_a - ids_f
    only_in_f = ids_f - ids_a
    print("course_uuid only in toorandmed_aasta.csv:", len(only_in_a))
    print(list(only_in_a)[:10])
    print("course_uuid only in toorandmed.csv:", len(only_in_f))
    print(list(only_in_f)[:10])
    print()

    # Compare values for shared course_uuid
    common_ids = ids_f & ids_a
    merged = df_full.merge(df_aasta, on='course_uuid', suffixes=('_full','_aasta'))
    diffs = []
    for col in common_cols:
        if col == 'course_uuid': 
            continue
        c_full = col + '_full'
        c_aasta = col + '_aasta'
        if c_full in merged.columns and c_aasta in merged.columns:
            neq = merged[merged[c_full].fillna('<<NA>>') != merged[c_aasta].fillna('<<NA>>')]
            if not neq.empty:
                diffs.append((col, len(neq)))
    print("Columns with differing values (column, differing_rows_count):")
    print(diffs)
    print()

    # Show sample diffs for up to 5 columns
    for col, _ in diffs[:5]:
        c_full = col + '_full'
        c_aasta = col + '_aasta'
        print("Sample diffs for column:", col)
        print(merged.loc[merged[c_full].fillna('<<NA>>') != merged[c_aasta].fillna('<<NA>>'),
                         ['course_uuid', c_full, c_aasta]].head(5).to_string(index=False))
        print()
else:
    print("course_uuid not present in both files; cannot do row-level key comparison.")
```
